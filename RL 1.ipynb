```python
def transition(state, action):
    '''
    Actions : 
    0 -> Left 
    1 -> Right
    2 -> Up 
    3 -> Down
    '''
    # left 
    # terminal state
    if state == 9 or state == 14:
        return state 
    # means moving left 
    if action == 0: 
        if state in [0,5,10,8]:
            newstate = state 
        else: 
            newstate = state - 1
    # try to finish moving right, up and down by your self ! 
    if action == 1:
        if state in [4,5]:
            newstate = state # since we are at the edge and can't move right
        else:
            newstate = state + 1
    # Up
    elif action == 2: 
        if state in [1,2,10,11,12,13]:
            newstate = state
        else:
            newstate = state + 5
    # Down
    else: 
        if state in [11, 12, 0,1,2,3,4]:
            newstate = state
        else:
            newstate = state -5
    return newstate
# the reward function for each move ! 
def reward(state, action, newstate):
    r = 0 # total reward
    # we have three conditions
    if newstate == 14: 
        r = 1
    elif newstate == 9: 
        r = -1
    else:
        r = 0
    return r 
def step(state, action):
    newstate = transition(state, action)
    r = reward(state, action, newstate)
    # episodes ends with a non-zero reward 
    done = (r!=0)
    return newstate, r, done
import random 
def sampleaction(): # used for the random policy 
    return random.randint(0,3)
def reset(): # for reset the environment 
    return 0, 0, False
def render(state):
    '''
    display the 3*5 grid word
    '''
    print("#" * 21)
    for i in range(2, -1, -1):
        row = "#"
        border = "#" if i == 0 else "-"

        for j in range(5):
            pos = i * 5 + j
            if pos == state:
                cell = " * "
            elif pos in [6,7]:
                cell = "###" 
            else: 
                cell = "   "
            if j < 4: 
                row += cell + "|"
            else: 
                row += cell 
        row += "#"
        print(row)
        print("#" + border * 19 + "#")
```

```python
def render(state):
    '''
    display the 3*5 grid word
    '''
    print("#" * 21)
    for i in range(2, -1, -1):
        row = "#"
        border = "#" if i == 0 else "-"

        for j in range(5):
            pos = i * 5 + j
            if pos == state:
                cell = " * "
            elif pos in [6,7]:
                cell = "###" 
            else: 
                cell = "   "
            if j < 4: 
                row += cell + "|"
            else: 
                row += cell 
        row += "#"
        print(row)
        print("#" + border * 19 + "#")
```

```python
num_runs = 10
stepcounter = 0
summedreward = 0 
for episode in range(num_runs):
    print(f"Episode {episode + 1}")
    state, r, done = reset()
    while not done:
        action = sampleaction() # sample a number from that strategy and move 
        newstate, r, done = step(state, action)
        stepcounter += 1
        state = newstate
        render(state)
        summedreward += r
        print()
print(f"Average number of step per rum: {stepcounter/num_runs:.2f}")
print(f"Average reward over {num_runs} runs: {summedreward}")
```
