```python
import numpy as np
import gymnasium as gym
import minigrid
import matplotlib.pyplot as plt
from BasicObsWrapper import SimpleObsWrapper

import random


def choose(q, state, epsilon):
    """
    epsilon-greedy action selection

    Parameters
    ----------
    q : np.array (states x actions)
        A q table for each action, state combination
    state : int (0..143)
        The current state of the agent.
    epsilon : float (between 0 and 1)
        Probability of taking a random action.

    Returns
    -------
    int
        The chosen action, representing one of 
        env.actions.left: 0, env.actions.right: 1, env.actions.forward: 2

    """
    if random.random() <= epsilon:
        action = random.randint(0, q.shape[1]-1)
    else:
        action = np.argmax(q[state,:])
        
    return action


def sarsa(env, alpha, epsilon, gamma,
            num_episodes = 50, num_states = 144, num_actions = 3):
      """
      SARSA for our minigrid environment
      Parameters
      ----------
      env : environment
          a gym environment
      alpha : float
          a learning rate.
      epsilon : float
          a probability for choosing a random action (0 <= epsilon <= 1).
      gamma : float
          a discount value (0 < gamma < 1).
      num_episodes : int, optional
          Number of episodes to train. The default is 50.
      num_states : int, optional
          Number of states in the environment. The default is 144.
      num_actions : int, optional
          Number of actions available. The default is 3.
         
      Returns
      -------
      q : array (num_states x num_actions)
          state-action value table.
          
      """
      q = 0.333 * np.ones((num_states, num_actions))
      while num_episodes > 0:          
          if num_episodes % 10 == 0:
              print(f'Episodes left: {num_episodes}')

          done = truncated = False 
          s0, info = env.reset() # start state
          a0 = choose(q, s0, epsilon)   

          while not done and not truncated:
              if num_episodes % 199 == 0:
                  img = env.render()
                  plt.close()
                  plt.imshow(img)
                  plt.show()
                  
              s1, reward, done, truncated, info  = env.step(a0)
              a1 = choose(q, s1, epsilon)
              q[s0,a0] = q[s0,a0] + alpha * (reward + gamma * q[s1,a1] - q[s0,a0])
              s0 = s1
              a0 = a1
             
          num_episodes = num_episodes - 1

      return q
 
# helper function for visualisation
# arrow characters are from https://en.wikipedia.org/wiki/Arrow_(symbol)
def vact(action, state):
    arrows = '↺↻→↺↻↓↺↻←↺↻↑'
    # looking right: 0, looking down: 1, looking left: 2, looking up: 3
    direction = state // 36
    return arrows[direction * 3 + action]

 # text-based policy visualisation   
def visualise(q):
    
    for row in range(6):
        for i in range(4):
            line = ''
            for col in range(6):
                if i == 0:
                    state = row*6 + col + 108
                    a = np.argmax(q[state,:])
                    line += ' ' + vact(a, state) + '   '
                elif i == 1:
                    state = row*6 + col + 72
                    a = np.argmax(q[state,:])                    
                    line += vact(a, state) + ' '
                    state = row*6 + col
                    a = np.argmax(q[state,:])                    
                    line += vact(a, state) + '  '
                elif i == 2:
                    state = row*6 + col + 36
                    a = np.argmax(q[state,:])                    
                    line += ' ' + vact(a, state) + '   '
                else:
                    line += '     '
                    
            print(line)
```

```python
import gymnasium as gym
import torch
from torch.autograd import Variable 
import matplotlib.pyplot as plt 
import copy
import numpy as np
import random

def plot_res(values, title='', save = False):   
    ''' Plot the reward curve and histogram of results over time.'''

    # Define the figure
    f, ax = plt.subplots(nrows=1, ncols=2, figsize=(12,5))
    f.suptitle(title)
    ax[0].plot(values, label='score per run')
    ax[0].axhline(195, c='red',ls='--', label='goal')
    ax[0].set_xlabel('Episodes')
    ax[0].set_ylabel('Reward')
    x = range(len(values))
    ax[0].legend()
    # Calculate the trend
    try:
        z = np.polyfit(x, values, 1)
        p = np.poly1d(z)
        ax[0].plot(x,p(x),"--", label='trend')
    except:
        pass
    
    # Plot the histogram of results
    ax[1].hist(values[-50:])
    ax[1].axvline(195, c='red', label='goal')
    ax[1].set_xlabel('Scores per Last 50 Episodes')
    ax[1].set_ylabel('Frequency')
    ax[1].legend()
    if save:
        plt.savefig('perf.png', dpi=300)
    plt.show()


def random_search(env, episodes, title='Random Strategy'):
    """ Random search strategy implementation."""
    final = []
    for episode in range(episodes):
        env.reset()
        done = truncated = False
        total = 0
        while not done and not truncated:
            # Sample random actions
            action = env.action_space.sample()
            # Take action and extract results
            next_state, reward, done, truncated, _ = env.step(action)
            # Update reward
            total += reward
            if done:
                break
        # Add to the final reward
        final.append(total)
        plot_res(final,title)
    return final

env = gym.envs.make("CartPole-v1")
# Number of states
n_state = env.observation_space.shape[0]
# Number of actions
n_action = env.action_space.n
# Number of episodes
episodes = 20

random_search(env, episodes)
```

```python
import gymnasium as gym
import torch
from torch.autograd import Variable
import matplotlib.pyplot as plt
import copy
import numpy as np
import torch.nn as nn
import torch.nn.functional as F
import random

class DQN(nn.Module):
    ''' Deep Q Neural Network class. '''
    def __init__(self, state_dim, action_dim, hidden_dim=64, lr=0.05):

        # add your code here
        # state_dim : is the number of possible state
        # action_dim: is the number of possible actions 
        super(DQN, self).__init__()

        self.fc1 = nn.Linear(state_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim * 2)
        self.fc3 = nn.Linear(hidden_dim * 2, action_dim)
        self.optimizer = torch.optim.Adam(self.parameters(), lr=lr)

    def forward(self, x):
        # x -> fc1 -> ReLU -> fc2 -? ReLU -> fc3 -> y
        x = F.leaky_relu(self.fc1(x))
        x = F.leaky_relu(self.fc2(x))
        x = self.fc3(x)
        return x

    def update(self, state, y):
        """Update the weights of the network given a training sample. """

        # add your code here
        # state ： state to use for the loss computation
        # y: the expectation 
        # backupdate  
        y_pred = self(torch.Tensor(state))
        criterion = nn.MSELoss()
        loss = criterion(y_pred, torch.Tensor(y))
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

    def predict(self, state):
        """ Compute Q values for all actions using the DQL. """

        # add your code here
        with torch.no_grad():
            return self(torch.Tensor(state))
```

```python
def q_learning(env, model, episodes, gamma=0.9, 
               epsilon=0.3, eps_decay=0.99,
               title = 'DQL', verbose=True):
    """Deep Q Learning algorithm using the DQN. """
    final = []
    for i, episode in enumerate(range(episodes), start = 1):        
        # Reset state
        state, _ = env.reset()
        done = truncated = False
        total = 0
        
        while not done and not truncated:
            ### Implement greedy search policy 
            if random.random() <= epsilon:
                action = env.action_space.sample()
            else:
                q_values = model.predict(state)
                action = torch.argmax(q_values).item()
            
            # Take action and add reward to total
            next_state, reward, done, truncated, _ = env.step(action)
            
            # Update total
            total += reward 
            q_values = model.predict(state).tolist()

            if done:
                q_values[action] = reward
                model.update(state, q_values)
                break
                # Update network weights
            q_values_next = model.predict(next_state)
            q_values[action] = reward + gamma * torch.max(q_values_next).item()
            model.update(state, q_values)
            state = next_state
            ### end of your code
        
        # Update epsilon
        epsilon = max(epsilon * eps_decay, 0.01)
        final.append(total) 
        plot_res(final, title, i == episodes)
        
        if verbose:
            print("episode: {}, epsilon: {}, total reward: {}".format(i, epsilon, total))
        
    return final 
```

```python
env = gym.envs.make("CartPole-v1")
n_state = env.observation_space.shape[0]
n_action = env.action_space.n
episodes = 250
n_hidden = 150
lr = 0.001
```

```python
dqn = DQN(n_state, n_action, n_hidden, lr)
q_learn = q_learning(env, dqn, episodes, gamma = 0.9, epsilon = 0.01)
```
