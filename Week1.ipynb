```python
# Challenge 2
# y = f(w.T*x) 
# w = [-1,2,3] # w is 1 * 3 
# x = [1,1,1]  # x is 1 * 3 
# First step: define the function f 
# second step : take w.T times x as input 
```

```python
# def is the key words for defining a function 
# the blue one sigmoid is the name of the function 
# (x) stands for the input
import math
def sigmoid(x):
    return 1/(1+math.exp(-x))
# neuron is f
def neuron(x, w):
    return(sigmoid(x.dot(w)))
x = np.array([1,1,1])
w = np.array([-1,2,3])
```

```python
neuron(x, w)
```
